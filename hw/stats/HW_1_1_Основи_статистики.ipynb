```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]

mean_visits = sum(visits) / n_sum_purchases
v_minus_mean = [(i - mean_visits)**2 for i in visits]
std_visits = pow(sum(v_minus_mean) / (n_sum_purchases - 1), 0.5)
```

```python
sv = [(sum_purchases[i] - mean_sum_purchases)*(visits[i] - mean_visits) for i in range(n_sum_purchases)]
cov_sv = sum(sv) / (n_sum_purchases - 1)

result = cov_sv / (std_sum_purchases * std_visits)

print(f"Covariance of sum_purchases and visits: {cov_sv:.2f}")
print(f"Coefficient Pearson of sum_purchases and visits: {result:.2f}")
```
